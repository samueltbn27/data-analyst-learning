## Join Tuples


```python
identitas = (
    101,
    "Samuel"
)

pekerjaan = (
    "Data Engineer",
    "Bandung"
)

skill = (
    "Python",
    "SQL",
    "Apache Airflow"
)

profil = identitas + pekerjaan + skill

print(f"Profil : {profil}")
print("Jumlah Elemen : ", len(profil))
print(f"Elemen Pertama : {profil[0]}")
print(f"Elemen Terakhir : {profil[6]}")
```

```python
pegawai = (
    (101, "Samuel"),
    (102, "Budi"),
    (103, "Sinta"),
    (104, "Andi"),
    (105, "Rina")
)

pekerjaan = (
    ("Data Engineer", 95),
    ("Data Analyst", 81),
    ("Backend Developer", 88),
    ("Data Scientist", 72),
    ("Business Analyst", 90)
)

jumlah_pegawai = 0
total_nilai = 0

# 1. Gabungkan kedua tuple menggunakan operator +
data = tuple(p1 + p2 for p1, p2 in zip(pegawai, pekerjaan))
for item in data:
    print(item)

print()
# 2. Loop
for index, item in enumerate(data, start=1):

    id_pegawai, nama, jabatan, nilai = item

    if nilai <= 80:
        continue

    if nama == "Rina":
        break

    jumlah_pegawai += 1
    total_nilai += nilai

    print(f"ID : {id_pegawai} | {nama} | {jabatan} | {nilai} ")

rata_rata = total_nilai / jumlah_pegawai

print()
print(f"Jumlah Pegawai : {jumlah_pegawai}")
print(f"Total Nilai ; {total_nilai}")
print(f"Rata Rata : {rata_rata}")

print()
print(f"Elemen Terakhir : {data[4][2:4]}")
print(f"Tiga Elemen Terakhir : {data[2:4]}")
```

```python
identitas = (
    (101, "Samuel"),
    (102, "Budi"),
    (103, "Sinta"),
    (104, "Andi"),
    (105, "Rina"),
    (106, "Dewi")
)

pekerjaan = (
    ("Data Engineer", "Bandung"),
    ("Data Analyst", "Jakarta"),
    ("Backend Developer", "Bandung"),
    ("Data Scientist", "Surabaya"),
    ("Business Analyst", "Bandung"),
    ("Machine Learning Engineer", "Jakarta")
)

performa = (
    (95, True),
    (82, True),
    (88, False),
    (76, True),
    (91, True),
    (98, True)
)

jumlah_pekerja_aktif = 0
total_nilai = 0
kota_bandung = 0

# Bagian 1 – Join Tuple
data_karyawan = tuple(i1 + p2 + p3 for i1, p2, p3 in zip(identitas, pekerjaan, performa))
for item in data_karyawan:
    print(item)

print()
# Bagian 2 – Loop
for index, (id_pekerja, nama, jabatan, kota, nilai, performa) in enumerate(data_karyawan, start=1):

    if performa == False:
        continue

    if nama == "Dewi":
        break

    if kota == "Bandung":
        kota_bandung += 1

    jumlah_pekerja_aktif += 1
    total_nilai += nilai
    print(f"{index}. ID : {id_pekerja} | {nama} | {jabatan} | {kota} | Nilai : {nilai}")



rata_rata = total_nilai / jumlah_pekerja_aktif
print()
print(f"Jumlah Pegawai Aktif : {jumlah_pekerja_aktif}")
print(f"Total Nilai : {total_nilai}")
print(f"Rata Rata : {rata_rata}")
print(f"Jumlah Kota Bandung : {kota_bandung}")
```
