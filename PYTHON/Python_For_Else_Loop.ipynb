```python
students = [
    {"nim": "231001", "name": "Andi", "gpa": 3.75},
    {"nim": "231002", "name": "Budi", "gpa": 4.20},
    {"nim": "231003", "name": "Citra", "gpa": 3.50},
    {"nim": "231004", "name": "", "gpa": 3.80}
]

for student in students:
    if student["nim"] == "" or student["name"] == "" or student["gpa"] < 0 or student["gpa"] > 4.00:
        print("Data mahasiswa tidak valid!")
        print(f"Nim : {student["nim"]} \nName : {student["name"]} \nGpa : {student["gpa"]}")
        break
else:
    print("Semua data mahasiswa valid.")
```

```python
transactions = [
    {"transaction_id": "TRX001", "customer": "Andi", "amount": 150000},
    {"transaction_id": "TRX002", "customer": "Budi", "amount": 250000},
    {"transaction_id": "", "customer": "Citra", "amount": 180000},
    {"transaction_id": "TRX004", "customer": "Dina", "amount": -50000},
    {"transaction_id": "TRX005", "customer": "Eka", "amount": 300000}
]

for index, transaction in enumerate(transactions, start=1):
    if (
        transaction["transaction_id"] == "" or
        transaction["customer"] == "" or
        transaction["amount"] <= 0
    ):
        if transaction["transaction_id"] == "":
            reason = "Transaction ID kosong"

        elif transaction["customer"] == "":
            reason = "Customer kosong"

        elif transaction["amount"] <= 0:
            reason = "Amount harus lebih dari 0"

        print(f"Data transaksi ke-{index} tidak valid!")
        print(f"Alasan        : {reason}")
        print(
            f"Transaction ID : {transaction["transaction_id"]}\n"
            f"Customer : {transaction["customer"]}\n"
            f"Amount : {transaction["amount"]}" 
        )
        break
else:
    print("Semua transaksi valid.")
```

```python
orders = [
    {
        "order_id": "ORD001",
        "customer": "Andi",
        "items": [
            {"product": "Laptop", "price": 12000000},
            {"product": "Mouse", "price": 250000}
        ]
    },
    {
        "order_id": "ORD002",
        "customer": "Budi",
        "items": [
            {"product": "Keyboard", "price": -500000}
        ]
    },
    {
        "order_id": "ORD003",
        "customer": "Citra",
        "items": [
            {"product": "Monitor", "price": 2500000}
        ]
    }
]

for order in orders:

    if order["order_id"] == "":
        print("Order tidak valid!")
        print("Alasan : Order ID kosong")
        print(f"Order ID : {order['order_id']}")
        print(f"Customer : {order['customer']}")
        break

    elif order["customer"] == "":
        print("Order tidak valid!")
        print("Alasan : Customer kosong")
        print(f"Order ID : {order['order_id']}")
        print(f"Customer : {order['customer']}")
        break

    for item in order["items"]:

        if item["price"] <= 0:
            print("Order tidak valid!")
            print("Alasan : Harga produk harus lebih dari 0")
            print(f"Order ID : {order['order_id']}")
            print(f"Customer : {order['customer']}")
            print(f"Produk : {item['product']}")
            print(f"Harga : {item['price']}")
            break
    else:
        # else milik for item
        continue
    # break ini akan dijalankan jika loop item berhenti karena break
    break
else:
    print("Semua order valid.")
```

```python
karyawan = [
    {"id": 1, "nama": "Andi", "departemen": "IT", "gaji": 8500000},
    {"id": 2, "nama": "Budi", "departemen": "HR", "gaji": 4500000},
    {"id": 3, "nama": "Citra", "departemen": "IT", "gaji": 12000000},
    {"id": 4, "nama": "Dina", "departemen": "Finance", "gaji": 7000000},
    {"id": 5, "nama": "Eko", "departemen": "IT", "gaji": -1000000},
    {"id": 6, "nama": "Fajar", "departemen": "HR", "gaji": 0},
    {"id": 7, "nama": "Gina", "departemen": "Finance", "gaji": 5000000}
]

Junior = 0
Senior = 0
Manager = 0 
Data_Tidak_Valid = 0
Total_Gaji_Valid =0
Jumlah_Data_Valid = 0

for i in karyawan:

    if i["gaji"] <= 0:
        status = "Data Tidak Valid"
    elif i["gaji"] < 5000000:
        status = "Junior"
    elif i["gaji"] <= 10000000:
        status = "Senior"
    else:
        status = "Manager"

    if status == "Junior":
        Junior += 1
    elif status == "Senior":
        Senior += 1
    elif status == "Manger":
        Manager += 1
    else:
        Data_Tidak_Valid += 1

    if i["gaji"] >= 0:
        Total_Gaji_Valid += i["gaji"]
        Jumlah_Data_Valid += 1
    
    if i["gaji"] >= 0:
        rata_rata = Total_Gaji_Valid / Jumlah_Data_Valid


    print(f"ID {i["id"]} | {i["nama"]} | {i["departemen"]} | {i["gaji"]} | {status}")

print()
print("===== Ringkasan Status =====")
print(
    f"Junior : {Junior}\n"
     f"Senior : {Senior}\n"
    f"Manager : {Manager}\n"
    f"Data Tidak Valid : {Data_Tidak_Valid}\n"
)

print("===== Gaji Valid =====")
print(
    f"Total Gaji Valid : {Total_Gaji_Valid:,}"
)

print()
print("===== Rata-rata Gaji  =====")
print(
    f"Rata Rata Gaji : {rata_rata:,}"
)  
```
