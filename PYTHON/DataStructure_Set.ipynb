## Access Set Items

```python
bahasa = {"Python", "SQL", "Java", "Scala"}

# 1. Menampilkan seluruh isi set
print("Daftar bahasa pemrograman:")
for item in bahasa:
    print(item)

# 2. Mengecek apakah "Python" ada
if "Python" in bahasa:
    print("\nPython ditemukan.")
else:
    print("\nPython tidak ditemukan.")

# 3. Mengecek apakah "R" ada
if "R" in bahasa:
    print("R ditemukan.")
else:
    print("R tidak ditemukan.")

# 4. Mengecek apakah "SQL" ada
if "SQL" in bahasa:
    print("SQL tersedia dalam set.")
else:
    print("SQL tidak tersedia dalam set.")
```

```python
id_karyawan = {1001, 1002, 1003, 1004, 1005}

print("Daftar ID Karyawan :")
for item in id_karyawan:
    print(item)

if 1003 in id_karyawan:
    print("\n1003 Ditemukan")
else:
    print("1003 Tidak Ditemukan")

if 1010 in id_karyawan:
    print("\n1010 Ditemukan")
else:
    print("1010 Tidak Ditemukan")

print(f"Jumlah Karyawan Aktif : ", len(id_karyawan))

if 1005 in id_karyawan:
    print("ID 1005 adalah karyawan aktif")
else:
    print("ID 1005 bukan karyawan aktif")
```

```python
tools = {"Python", "SQL", "Git"}

print("Daftar Tools Awal :")
for item in tools:
    print(item)

print("\nDaftar Tools Setelah Ditambahkan:")
tools.add("Docker")
tools.update(["Apache Airflow", "PostgreSQL", "Python"])
for item in tools:
    print(item)

print("\nJumlah Tools :", len(tools))
```

## Join Sets

```python
tim_a = {"Python", "SQL", "Git"}

tim_b = {"Git", "Docker", "Apache Airflow"}

print("Tools Tim A :")
for item in tim_a:
    print(item)

print("\nTools Tim B :")
for item in tim_b:
    print(item)

print("\nSemua Tools :")
semua_tools = tim_a.union(tim_b)
for item in semua_tools:
    print(item)

print("\nJumlah Tools : ", len(semua_tools))
```

```python
pelanggan_a = {101, 102, 103, 104}

pelanggan_b = {103, 104, 105, 106}

# Tampilkan isi pelanggan_a menggunakan loop.
print("Pelanggan A :")
for item in pelanggan_a:
    print(item)

# Tampilkan isi pelanggan_b menggunakan loop.
print("\nPelanggan B : ")
for item in pelanggan_b:
    print(item)

# Tambahkan pelanggan baru
pelanggan_a.update([107, 108])

# Hapus pelanggan yang sudah tidak aktif
pelanggan_a.remove(102)

semua_pelanggan = pelanggan_a.union(pelanggan_b)

print("\nSemua Pelanggan :")
for item in semua_pelanggan:
    print(item)

print("Jumlah pelanggan unik :", len(semua_pelanggan))
```

```python
cabang_bandung = {101, 102, 103, 104}

cabang_jakarta = {103, 104, 105, 106}

cabang_surabaya = {106, 107, 108}

print("Cabang Bandung :")
for item in cabang_bandung:
    print(item)

print("\nCabang Jakarta :")
for item in cabang_jakarta:
    print(item)

print("\nCabang Surabaya :")
for item in cabang_surabaya:
    print(item)

cabang_bandung.update([109, 110])

cabang_bandung.remove(102)
cabang_surabaya.remove(108)

semua_pelanggan = cabang_bandung.union(cabang_jakarta, cabang_surabaya)

print("\nSemua Pelanggan :")
for item in semua_pelanggan:
    print(item)

print("\nJumlah Pelanggan :", len(semua_pelanggan))

target_pelanggan = {105, 102, 110}
for id_p in target_pelanggan:
    if id_p in semua_pelanggan:
        print(f"ID {id_p} ditemukan")
    else:
        print(f"ID {id_p} tidak ditemukan")
        
pelanggan_list = list(semua_pelanggan)
for index, item in enumerate(pelanggan_list, start= 1):
    print(f"Pelanggan ke-{index} : {item}")
```
