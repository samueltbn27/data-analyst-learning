# Python If-Else dan Match Case

## 1. If-Else One Line (Ternary Operator)

Bentuk Umum

- nilai_jika_true if kondisi else nilai_jika_false

```python
# Contoh Sederhana

umur = 20

if umur >= 18:
    print("Dewasa")
else:
    print("Anak Anak")
```

```python
# One Line
umur = 20

status = "Dewasa" if umur >= 18 else "Anak anak"

print(status)
```

```python
# Contoh Sederhana Jika Menggunakan Elif
umur = 20

if umur >= 18:
    print("Dewasa")
elif umur >= 15:
    print("Remaja")
else:
    print("Anak Anak")
```

```python
# One Line
umur = 15

status = "Dewasa" if umur >= 18 else "Remaja" if umur >= 15 else "Anak Anak"

print(status)
```

## 2. Match-Case

melakukan pencocokan pola (structural pattern matching).

### 2.1 Contoh Sederhana

```python
hari = "Senin"

match hari:
    case "Senin":
        print("Awal Minggu")
    case "Jumat":
        print("Akhir Kerja")
    case _:
        print("Hari Biasa")
```

### 2.2 Contoh Menentukan Jenis File yang Masuk ke Pipeline.

```python
file_type = "csv"

match file_type:
    case "csv":
        print("Load CSV")
    case "json":
        print("Load JSON")
    case "parquet":
        print("Load Parquet")
    case _:
        print("Format tidak didukung")
```

```python
job_status = "FAILED"

match job_status:
    case "PENDING":
        print("Menunggu Proses ETL")
    case "RUNNING":
        print("ETL Sedang Berjalan")
    case "SUCCESS":
        print("ETL Berhasil")
    case "FAILED" | "CANCELLED":
        print("Perlu Investigasi Tim Data")
    case _:
        print("Status Tidak Diketahui")
```

```python
pipeline = {
    "status": "SUCCESS",
    "records": 250000
}

match pipeline:
    case {"status":"SUCCESS","records":x} if x >= 100000:
        print("Data Warehouse Berhasil Diupdate")
    case {"status":"SUCCESS","records":x} if x < 100000:
        print("Update Berhasil dengan Data Kecil")
    case {"status":"FAILED"}:
        print("Jalankan Ulang Pipeline")
    case {"status":"RUNNING"}:
        print("Pipeline Sedang Berjalan")
    case _:
        print("Status Tidak Dikenal")
```
