## Access Dictionary Items

```python
karyawan = {
    "id": 101,
    "nama": "Samuel",
    "jabatan": "Data Analyst",
    "gaji": 8500000,
    "kota": "Bandung"
}

print(f"Nama : {karyawan['nama']}")
print(f"Jabatan : {karyawan['jabatan']}")
print(f"Gaji : Rp{karyawan['gaji']:,}".replace(",", "."))
print(f"Kota : {karyawan.get('kota')}")

email_pegawai = karyawan.get("email", "Email Tidak Tersedia")
print(email_pegawai)
```

```python
karyawan = {
    "id": 101,
    "nama": "Samuel",
    "departemen": "Data Analytics",
    "gaji_pokok": 8500000,
    "bonus": 1500000,
    "alamat": {
        "kota": "Bandung",
        "provinsi": "Jawa Barat"
    },
    "skill": [
        "Python",
        "SQL",
        "Power BI",
        "Excel"
    ]
}

print(f"Nama : {karyawan["nama"]}")
print(f"Departemen : {karyawan["departemen"]}")
total_penghasilan = karyawan["gaji_pokok"] + karyawan["bonus"]
print(f"Total Penghasilan : Rp{total_penghasilan:,}".replace(",","."))
print(f"Kota : {karyawan["alamat"]["kota"]}")

skill_pertama = karyawan["skill"][0]
skill_terakhir = karyawan["skill"][-1]
print(f"Skill Pertama : {skill_pertama}")
print(f"Skill Terakhir : {skill_terakhir}")

email_karyawan = karyawan.get("email", "Belum tersedia")
print(f"Email : {email_karyawan}")

print("Jumlah Karyawan :", len(karyawan["skill"]))

print("\nData Karyawan")
for key, value in karyawan.items():
    if key != "alamat" and key != "skill":
        print(f"{key} : {value}")

print(f"\nProvinsi : {karyawan["alamat"]["provinsi"]}")
```
