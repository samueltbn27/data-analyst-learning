# If Elif Else Statements

## 1. If Statement di Python

if digunakan untuk menjalankan kode hanya jika suatu kondisi bernilai True.

### 1.1 Contoh 1

```python
umur = 18

if umur >= 17:
    print("Anda bisa membuat SIM")
```

Latihan

```python
usia = 18

if usia >= 17:
    print("Mendapat kartu anggota")
```

## 2. Else Statement di Python

else digunakan untuk menjalankan kode ketika kondisi pada if bernilai False.

- Jika kondisi benar (True), jalankan if.
- Jika kondisi salah (False), jalankan else.

### 2.1 Contoh 1

```python
umur = 15

if umur >= 17:
    print("Boleh membuat SIM")
else:
    print("Belum boleh membuat SIM")
```

Latihan

```python
umur = 10

if umur >= 13:
    print("Boleh menonton")
else:
    print("Tidak boleh menonton")
```

## 3. Elif Statement

elif digunakan ketika ada lebih dari dua kemungkinan kondisi.

### 3.1 Contoh 1

```python
nilai = 85

if nilai >= 90:
    print("A")
elif nilai >= 75:
    print("B")
else:
    print("C")
```

Latihan

```python
nilai = 78
kehadiran = 65

if nilai >= 90 and kehadiran >= 80:
    print("Lulus dengan Predikat A")
elif nilai >= 75 and kehadiran >= 70:
    print("Lulus")
elif nilai >= 60:
    print("Remedial")
else:
    print("Tidak Lulus")
```

## 4. Nested If di Python

digunakan ketika kita ingin membuat keputusan yang lebih detail setelah kondisi pertama terpenuhi.

### 4.1 Contoh 1

```python
umur = 20
punya_ktp = True

if umur >= 17:
    if punya_ktp:
        print("Boleh masuk")
    else:
        print("Harus punya KTP")
else:
    print("Tidak cukup umur")
```

Latihan

```python
status_login = True
role = "user"

if status_login == True:
    if role == "admin":
        print("Akses penuh")
    elif role == "user":
        print("Akses terbatas")
    else:
        print("Role tidak dikenal")
else:
    print("Silakan login dulu")
```

## 5. Logical Operators di Python

Logical operators digunakan untuk menggabungkan atau membalik beberapa kondisi dalam if.

🧠 Jenis Logical Operators
| Operator | Nama   | Fungsi                                            |
| -------- | ------ | ------------------------------------------------- |
| `and`    | DAN    | True jika **dua kondisi True**                    |
| `or`     | ATAU   | True jika **salah satu kondisi True**             |
| `not`    | NEGASI | Membalik nilai (True jadi False, False jadi True) |


### 5.1 and (DAN)

```python
nilai = 80
absen = 90

if nilai >= 75 and absen >= 80:
    print("Lulus")
```

### 5.2 or (ATAU)

```python
nilai = 60

if nilai >= 75 or nilai >= 50:
    print("Lulus")
```

### 5.3 not (NEGASI)

```python
lulus = True

if not lulus:
    print("Remedial")
else:
    print("Lulus")
```

Latihan

```python
nilai = 82
kehadiran = 85

if nilai >= 85 and kehadiran >= 80:
    print("Beasiswa Penuh")
elif nilai >= 75 and kehadiran >= 75:
    print("Beasiswa Sebagian")
else:
    print("Tidak mendapat beasiswa")
```

## 6. Independent If di Python

Independent If adalah beberapa if yang berdiri sendiri dan tidak saling terhubung.

### 6.1 Contoh 1

```python
transaksi = 150000
login_harian = True
invite_teman = False

if transaksi >= 100000:
    print("Cashback 10,000")

if login_harian == True:
    print("Point Harian")

if invite_teman == True:
    print("Bonus Referal")
```

```python
email = ""
valid = True

email = email.strip()

if email == "":
    print("Email Tidak Valid Karena Kosong")
    email = False

if not ("." in email and "@" in email ):
    print("Email Wajib Mempunyai . dan @")
    email = False

if email.count("@") != 1:
    print("Email Hanya Boleh Mempunyai 1 @")
    email = False

if not email.endswith((".com", ".org", ".net")):
    print("Email Harus Ada Domain")
    email = False

if len(email) > 254:
    print("Standar email maksimal ±254 karakter")
    email = False

if not (email[0].isalnum() and email[-1].isalnum()):
    print("Karakter pertama dan terakhir harus huruf/angka")
    email = False

if valid:
    print("Email Valid")
```

```python
email = "@samuel@gmail.com"
valid = True

email = email.strip()

if email == "":
    print("Email tidak boleh kosong.")
    valid = False

if not ("@" in email and "." in email):
    print("Harus mengandung @ dan .")
    valid = False

if email.count("@") != 1:
    print("Email valid karena hanya memiliki satu @")
    valid = True

if not (email[0].isalnum() and email[-1].isalnum()):
    print("Email tidak valid karena karakter pertama bukan huruf atau angka")
    valid = False

if len(email) > 254:
    print("Email tidak valid karena panjang email melebihi 254 karakter")
    valid = False

if valid:
    print("Email Valid")
```

```python
email = "Baraa123@gmail.com"
password = "Baraa123"
valid = True

if password == "":
    print("Tidak boleh kosong")
    valid = False

if len(password) < 8:
    print("Minimal 8 karakter")
    valid = False

if password == password.lower():
    print("Harus ada minimal 1 huruf besar (uppercase)")
    valid = False

if password == password.upper():
    print("Harus ada minimal 1 huruf kecil (lowercase)")
    valid = False

if password == email:
    print("Tidak boleh sama dengan email")
    valid = False

if " " in password:
    print("Tidak boleh mengandung spasi")
    valid = False

if not (password[0].isalnum() and password[-1].isalnum()):
    print("Harus dimulai dan diakhiri dengan huruf atau angka")
    valid = False

if valid:
    print("Password Valid")
```
